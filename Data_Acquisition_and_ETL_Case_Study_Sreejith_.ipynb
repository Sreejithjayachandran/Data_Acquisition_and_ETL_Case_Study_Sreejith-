```python
#Task 6

df_gdp=df_gdp.rename(columns={"Country":"country"})
df_gdp

```

```python
#Task 7

df_population["population"] = pd.to_numeric(df_population["population"])
df_gdp["GDP"] = pd.to_numeric(df_gdp["GDP"])
df_internet["internet_users"] = pd.to_numeric(df_internet["internet_users"])
df_literacy["literacy_rate"] = pd.to_numeric(df_literacy["literacy_rate"])
print(df_population.dtypes)
print(df_gdp.dtypes)
print(df_internet.dtypes)
print(df_literacy.dtypes)
```

```python
#Task 8
merged_df = (
    df_population
    .merge(df_gdp, on="country", how="inner")
    .merge(df_internet, on="country", how="inner")
    .merge(df_literacy, left_on="country", right_on="name", how="inner")
    .drop(columns=["name"])
)

print(merged_df.head())
```

```python
#Task 9

merged_df["Internet Penetration (%)"] = (
    merged_df["internet_users"] / merged_df["population"] * 100
).round(2)

merged_df["GDP Per Capita"] = (
    merged_df["GDP"] * 1_000_000 / merged_df["population"]
).round(2)

print(merged_df.head())
```

```python
#Task 10

import sqlite3
import pandas as pd

# 1. Ensure columns are numeric
df_population["population"] = pd.to_numeric(df_population["population"])
df_gdp["GDP"] = pd.to_numeric(df_gdp["GDP"])
df_internet["internet_users"] = pd.to_numeric(df_internet["internet_users"])
df_literacy["literacy_rate"] = pd.to_numeric(df_literacy["literacy_rate"])

# 2. Merge all datasets using the correct keys
merged_df = (
    df_population
    .merge(df_gdp, on="country", how="inner")
    .merge(df_internet, on="country", how="inner")
    .merge(df_literacy, left_on="country", right_on="name", how="inner")
    .drop(columns=["name"])
)

# 3. Add engineered features
merged_df["Internet Penetration (%)"] = (
    merged_df["internet_users"] / merged_df["population"] * 100
).round(2)

merged_df["GDP Per Capita"] = (
    merged_df["GDP"] * 1_000_000 / merged_df["population"]
).round(2)

print("Merged DataFrame Preview:")
print(merged_df.head())

# 4. Save to SQLite database
conn = sqlite3.connect("country_statistics.db")
merged_df.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)
conn.close()
```
